```python
def clean_columns(df):
    df = df.copy()
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
        .str.replace("\n", " ", regex=False)
        .str.replace(r"\s+", "_", regex=True)
    )
    return df

df_finance = clean_columns(df_finance)
df_training = clean_columns(df_training)
df_cold = clean_columns(df_cold)
df_abs = clean_columns(df_abs)

print("Colonnes finance :", df_finance.columns.tolist())
print("Colonnes training :", df_training.columns.tolist())
print("Colonnes cold :", df_cold.columns.tolist())
print("Colonnes abs :", df_abs.columns.tolist())
```

```python
# =========================
# Préparation des dimensions de filtrage
# Objectif : conserver year / entity / direction dans toutes les tables KPI
# =========================

LEGAL_ENTITY_COL = "entity"      # entité légale dans le fichier Training
DIRECTION_COL = "direction"      # direction dans le fichier Training
ABS_ENTITY_COL = "société"       # entité légale dans le fichier Absence
ABS_DIRECTION_COL = "organisation_1"  # à ajuster si une autre colonne représente mieux la direction


def normalize_employee_code(s):
    """Normalise les matricules pour faciliter les jointures entre fichiers."""
    return s.astype(str).str.strip().str.replace(r"\.0$", "", regex=True)


def add_missing_dimension_columns(df, entity_col="entity", direction_col="direction"):
    """Garantit que les colonnes de filtre existent, même si la source ne les contient pas."""
    df = df.copy()
    if entity_col not in df.columns:
        df[entity_col] = "Non renseigné"
    if direction_col not in df.columns:
        df[direction_col] = "Non renseigné"
    df[entity_col] = df[entity_col].fillna("Non renseigné").astype(str).str.strip()
    df[direction_col] = df[direction_col].fillna("Non renseigné").astype(str).str.strip()
    return df

# Training : source principale pour entity / direction
df_training = add_missing_dimension_columns(df_training, LEGAL_ENTITY_COL, DIRECTION_COL)
df_training["employee_code"] = normalize_employee_code(df_training["employee_code"])
df_training["seesion_start_date"] = pd.to_datetime(df_training["seesion_start_date"], errors="coerce")
df_training["year"] = df_training["seesion_start_date"].dt.year

# Cold review : enrichissement entity / direction à partir du fichier Training
if "matricule" in df_cold.columns:
    df_cold["employee_code"] = normalize_employee_code(df_cold["matricule"])
else:
    df_cold["employee_code"] = np.nan

cold_date_col = "date_de_début_de_session" if "date_de_début_de_session" in df_cold.columns else "date"
df_cold[cold_date_col] = pd.to_datetime(df_cold[cold_date_col], errors="coerce")
df_cold["year"] = df_cold[cold_date_col].dt.year

training_dimensions = (
    df_training
    .sort_values("seesion_start_date")
    .drop_duplicates("employee_code", keep="last")
    [["employee_code", LEGAL_ENTITY_COL, DIRECTION_COL]]
)

df_cold = df_cold.merge(training_dimensions, on="employee_code", how="left")
df_cold = add_missing_dimension_columns(df_cold, LEGAL_ENTITY_COL, DIRECTION_COL)

# Absences : renommage pour harmoniser les dimensions
if "employee_code" in df_abs.columns:
    df_abs["employee_code"] = normalize_employee_code(df_abs["employee_code"])

df_abs["date_absence"] = pd.to_datetime(df_abs["date_absence"], errors="coerce")
df_abs["year"] = df_abs["date_absence"].dt.year

df_abs[LEGAL_ENTITY_COL] = df_abs[ABS_ENTITY_COL] if ABS_ENTITY_COL in df_abs.columns else "Non renseigné"
df_abs[DIRECTION_COL] = df_abs[ABS_DIRECTION_COL] if ABS_DIRECTION_COL in df_abs.columns else "Non renseigné"
df_abs = add_missing_dimension_columns(df_abs, LEGAL_ENTITY_COL, DIRECTION_COL)

print("Dimensions disponibles :")
print("Training:", [c for c in ["year", LEGAL_ENTITY_COL, DIRECTION_COL] if c in df_training.columns])
print("Cold review:", [c for c in ["year", LEGAL_ENTITY_COL, DIRECTION_COL] if c in df_cold.columns])
print("Absence:", [c for c in ["year", LEGAL_ENTITY_COL, DIRECTION_COL] if c in df_abs.columns])
```

```python
# =========================
# Fonctions communes pour sortir les KPI selon les 3 niveaux demandés
# =========================

DIMENSIONS = {
    "yearly": ["year"],
    "by_entity": ["year", "entity"],
    "by_direction": ["year", "direction"],
}


def build_kpi_views(df, kpi_name, value_col, agg="mean", count_col=None, value_name="value"):
    """
    Renvoie 3 vues pour un KPI :
    - valeurs par année
    - répartition par entité légale
    - répartition par direction
    """
    views = {}
    for view_name, group_cols in DIMENSIONS.items():
        grouped = (
            df
            .dropna(subset=["year"])
            .groupby(group_cols, dropna=False)
            .agg(
                value=(value_col, agg),
                n=(count_col or value_col, "count")
            )
            .reset_index()
        )
        grouped.insert(0, "KPI", kpi_name)
        grouped = grouped.rename(columns={"value": value_name})
        if pd.api.types.is_numeric_dtype(grouped[value_name]):
            grouped[value_name] = grouped[value_name].round(4)
        views[view_name] = grouped
    return views


def concat_kpi_views(kpi_views_dict, view_name):
    """Concatène la même vue pour plusieurs KPI afin d'obtenir une table dashboard-friendly."""
    return pd.concat([views[view_name] for views in kpi_views_dict.values()], ignore_index=True)
```

```python
# =========================
# KPI 1 — HCVA
# Human Capital Value Added
# =========================

pnb_row = df_finance[
    df_finance["toutes_filiales_conso"]
    .astype(str)
    .str.contains("Net Banking Income|PNB", case=False, na=False)
]

pnb_values = pnb_row[
    [
        "réel_décembre_2022",
        "réel_décembre_2023",
        "réel_décembre_2024",
        "réel_décembre_2025",
    ]
].iloc[0]

hcva_base = pd.DataFrame({
    "year": [2022, 2023, 2024, 2025],
    "pnb": pnb_values.values,
    "fte": [954.6, 926.3, 967.4, 6631.47],
})

hcva_base["HCVA"] = (hcva_base["pnb"] / hcva_base["fte"]).round(2)

# Le fichier Finance ne contient pas entity/direction au même niveau de granularité.
# On garde donc des valeurs globales filtrables via une modalité unique.
hcva_base["entity"] = "Toutes entités"
hcva_base["direction"] = "Toutes directions"

hcva_views = build_kpi_views(
    hcva_base,
    kpi_name="HCVA",
    value_col="HCVA",
    agg="mean",
    count_col="HCVA",
    value_name="HCVA"
)

hcva_views["yearly"]
```

```python
# =========================
# KPI 2 — KTI
# Knowledge Transfer Index
# =========================

application_col = "utilisez-vous_les_connaissances_acquises_lors_de_la_formation_?"

score_map = {
    "Oui, tout à fait": 1.0,
    "Oui, en partie": 0.5,
    "Non": 0.0,
    "Non, pas du tout": 0.0,
}

df_cold["application_score"] = df_cold[application_col].map(score_map)

kti_views = build_kpi_views(
    df_cold,
    kpi_name="KTI",
    value_col="application_score",
    agg="mean",
    count_col="application_score",
    value_name="KTI"
)

kti_views["by_entity"].head(10)
```

```python
# =========================
# KPI 3 — Skill Decay
# Collaborateurs sans formation depuis plus de 18 mois
# =========================

last_training = (
    df_training
    .sort_values("seesion_start_date")
    .dropna(subset=["employee_code"])
    .drop_duplicates("employee_code", keep="last")
    [["employee_code", "seesion_start_date", "year", "entity", "direction"]]
)

today = pd.Timestamp.today().normalize()
last_training["months_since_training"] = (
    (today - last_training["seesion_start_date"]).dt.days / 30.44
)
last_training["skill_decay_flag"] = last_training["months_since_training"] > 18

skill_decay_views = build_kpi_views(
    last_training,
    kpi_name="Skill Decay",
    value_col="skill_decay_flag",
    agg="mean",
    count_col="employee_code",
    value_name="pct_decay"
)

skill_decay_views["by_direction"].head(10)
```

```python
# =========================
# KPI 4 — RE-Score
# Engagement / taux d'absentéisme
# =========================

# 1. Filtrer vraies absences, exclure les congés
has_absence_group = "regroupement_jour_absences" in df_abs.columns
if has_absence_group:
    df_abs_filtered = df_abs[
        ~df_abs["regroupement_jour_absences"]
        .astype(str)
        .str.lower()
        .str.contains("cong", na=False)
    ].copy()
else:
    df_abs_filtered = df_abs.copy()

# 2. Scores engagement fixes, à remplacer par des scores par année / entité / direction si disponibles
ier_score = 70
imr_score = 77
trust_score = 78
engagement_score = np.mean([ier_score, imr_score, trust_score])
working_days = 218


def compute_re_score(df, group_cols):
    out = (
        df
        .dropna(subset=["year"])
        .groupby(group_cols, dropna=False)
        .agg(
            total_absence_days=("jours_ouvrés_absence", "sum"),
            nb_employees=("employee_code", "nunique")
        )
        .reset_index()
    )
    out["absence_rate"] = out["total_absence_days"] / (out["nb_employees"] * working_days)
    out["engagement_score"] = engagement_score
    out["RE_score"] = out["engagement_score"] / out["absence_rate"]
    out.insert(0, "KPI", "RE-Score")
    return out.round({"absence_rate": 4, "engagement_score": 2, "RE_score": 2})

re_views = {
    view_name: compute_re_score(df_abs_filtered, group_cols)
    for view_name, group_cols in DIMENSIONS.items()
}

re_views["yearly"]
```

```python
# =========================
# KPI 5 — SPE
# Strategic Payroll Elasticity / part des heures stratégiques
# =========================

strategic_keywords = [
    "data", "digital", "it", "technology",
    "cyber", "security", "risk", "compliance",
    "esg", "innovation", "transformation",
]


def classify_strategic(row):
    text = (
        str(row.get("attended_courses", "")) + " " +
        str(row.get("organization", "")) + " " +
        str(row.get("direction", ""))
    ).lower()
    return "Strategic" if any(k in text for k in strategic_keywords) else "Legacy"


df_training["category"] = df_training.apply(classify_strategic, axis=1)
df_training["strategic_hours"] = np.where(
    df_training["category"] == "Strategic",
    df_training["total_training_hours"],
    0
)


def compute_spe(df, group_cols):
    out = (
        df
        .dropna(subset=["year"])
        .groupby(group_cols, dropna=False)
        .agg(
            strategic_hours=("strategic_hours", "sum"),
            total_hours=("total_training_hours", "sum"),
            nb_formations=("employee_code", "count")
        )
        .reset_index()
    )
    out["strategic_share"] = out["strategic_hours"] / out["total_hours"]
    out.insert(0, "KPI", "SPE Strategic Share")
    return out.round({"strategic_share": 4})

spe_views = {
    view_name: compute_spe(df_training, group_cols)
    for view_name, group_cols in DIMENSIONS.items()
}

spe_views["by_entity"].head(10)
```

```python
# =========================
# Tables finales pour dashboard / export
# =========================

all_kpi_views = {
    "HCVA": hcva_views,
    "KTI": kti_views,
    "Skill Decay": skill_decay_views,
}

kpi_yearly = pd.concat(
    [
        hcva_views["yearly"],
        kti_views["yearly"],
        skill_decay_views["yearly"],
        re_views["yearly"],
        spe_views["yearly"],
    ],
    ignore_index=True,
    sort=False,
)

kpi_by_entity = pd.concat(
    [
        hcva_views["by_entity"],
        kti_views["by_entity"],
        skill_decay_views["by_entity"],
        re_views["by_entity"],
        spe_views["by_entity"],
    ],
    ignore_index=True,
    sort=False,
)

kpi_by_direction = pd.concat(
    [
        hcva_views["by_direction"],
        kti_views["by_direction"],
        skill_decay_views["by_direction"],
        re_views["by_direction"],
        spe_views["by_direction"],
    ],
    ignore_index=True,
    sort=False,
)

print("kpi_yearly", kpi_yearly.shape)
print("kpi_by_entity", kpi_by_entity.shape)
print("kpi_by_direction", kpi_by_direction.shape)

display(kpi_yearly.head(20))
display(kpi_by_entity.head(20))
display(kpi_by_direction.head(20))
```

```python
# Optionnel : export des tables finales
output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)

kpi_yearly.to_csv(output_dir / "kpi_yearly.csv", index=False)
kpi_by_entity.to_csv(output_dir / "kpi_by_entity.csv", index=False)
kpi_by_direction.to_csv(output_dir / "kpi_by_direction.csv", index=False)

print("Exports créés dans", output_dir.resolve())
```

### Résultat attendu

Le préprocessing conserve maintenant les dimensions nécessaires au filtrage :

- `year` pour suivre chaque KPI dans le temps ;
- `entity` pour filtrer / répartir par entité légale ;
- `direction` pour filtrer / répartir par direction.

Les trois tables finales à utiliser dans un dashboard sont :

- `kpi_yearly` : valeurs par année ;
- `kpi_by_entity` : valeurs par année et entité légale ;
- `kpi_by_direction` : valeurs par année et direction.

À noter : le fichier Finance utilisé pour le HCVA ne contient pas de granularité `entity` / `direction` exploitable dans ce notebook. Le HCVA est donc conservé en global avec les modalités `Toutes entités` et `Toutes directions`. Pour obtenir un vrai HCVA par entité ou direction, il faut une source Finance/FTE au même niveau de granularité.
